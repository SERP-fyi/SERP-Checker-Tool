```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/serp_benchmarks.csv')
print('Dataset shape:', df.shape)
print(df.head(10))
```

```python
intent_counts = df['intent'].value_counts()
colors = ['#3b82f6', '#10b981', '#f59e0b', '#8b5cf6']
plt.figure(figsize=(8, 5))
plt.pie(intent_counts, labels=intent_counts.index, colors=colors, autopct='%1.0f%%')
plt.title('SERPChecker — Intent Distribution', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
intent_avg = df.groupby('intent')['serp_feature_score'].mean().reset_index()
plt.figure(figsize=(8, 5))
plt.bar(intent_avg['intent'], intent_avg['serp_feature_score'], color=['#3b82f6', '#10b981', '#f59e0b', '#8b5cf6'])
plt.title('SERPChecker — Avg SERP Feature Score by Intent', fontsize=14, fontweight='bold')
plt.xlabel('Search Intent')
plt.ylabel('SERP Feature Score (0-100)')
plt.tight_layout()
plt.show()
print(intent_avg.to_string(index=False))
```
